```python
import numpy as np
import pandas as pd
import re

messages = pd.read_csv('SMSSpamCollection.csv', header=None, names=['raw'])
messages[['label', 'message']] = messages['raw'].str.split('\t', n=1, expand=True)
messages = messages.drop(columns='raw')
```

```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()
```

```python
corpus = []
for i in range(0, len(messages)):
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    review = review.lower()
    review = review.split()
    review = [lemmatizer.lemmatize(word) for word in review]
    review = ' '.join(review)
    corpus.append(review)
```

```python
corpus # word net lemmatier corpus
```

```python
from gensim.utils import simple_preprocess # converts docs into lowercase
```

```python
# one token list per message, so rows stay aligned with labels
words = []
for doc in corpus:
    words.append(simple_preprocess(doc))
```

```python
words
```

```python
## Train Word2Vec from scratch
import gensim

model = gensim.models.Word2Vec(words, vector_size=100, window=5, min_count=2, epochs=20)
```

```python
def avg_word2vec(doc):
    # remove out-of-vocabulary words
    vecs = [model.wv[word] for word in doc if word in model.wv]
    # messages with no known words get a zero vector instead of NaN
    if not vecs:
        return np.zeros(model.vector_size)
    return np.mean(vecs, axis=0)
```
