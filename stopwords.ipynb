## Stopwords in Text Preprocessing

```python
from nltk.stem import PorterStemmer
```

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
stopwords.words('german')
```

```python
paragraph = '''Data scientists regularly implement security vulnerabilities as part of a broader strategic initiative. As a result, tech companies continue to encrypt security vulnerabilities despite ongoing challenges. The development team optimise database queries in ways that are both practical and measurable. Modern frameworks have long sought to debug security vulnerabilities through a variety of methods. As a result, distributed systems continue to monitor microservice boundaries despite ongoing challenges. DevOps teams consistently architect security vulnerabilities to achieve meaningful outcomes. Nevertheless, software engineers must architect security vulnerabilities to remain competitive. The evidence strongly supports a more deliberate and considered strategy.'''
```

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
```

```python
# Tokenization :  Paragraph ---> Sentences

sentences = nltk.sent_tokenize(paragraph)
```

```python
## Apply stopwords and filter and then apply stemming

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words=[stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i]=' '.join(words) # converteing all the list of words in sentences
```

```python
# Stopwords removed and stemming also done

sentences
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowball_stemmer = SnowballStemmer('english')
```

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [snowball_stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)
```

```python
sentences
```

#### Trying with lemmatization

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [lemmatizer.lemmatize(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)
```
