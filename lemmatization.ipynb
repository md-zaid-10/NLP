## Lemmatization

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
'''
POS - 

Noun(n)-default
Verb(v)
adjective(a)
adverb(r)
'''

print('going ----> Verb :',lemmatizer.lemmatize('going', pos='v'))

print('going ----> noun :',lemmatizer.lemmatize('going', pos='n'))

print('going ----> Adjective :',lemmatizer.lemmatize('going', pos='a'))

print('going ----> Adverb :',lemmatizer.lemmatize('going', pos='r'))
```

```python
words = ['eating', 'eats', 'eaten', 'writing', 'writes', 'programming', 'programs', 'history', 'finally', 'finalize']

for word in words:
    print(word + '--->' + lemmatizer.lemmatize(word, pos='v'))
```
