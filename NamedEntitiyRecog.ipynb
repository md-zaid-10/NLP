## Named Entity Recognition

```python
sentence = 'The Eiffel Tower was built from 1887 to 1889 by French Engineer Gustave Eiffel, whose company specialiezd in building metal frameworks and structures.'

'''
NER Examples
Person : Napolean Bonaparte
Place : Paris
date : 10.02.2004
Time : 4:20 am
Money : 1 Million
Organization : United Nations
Percent : 45%
'''
```

```python
import nltk
words = nltk.word_tokenize(sentence)
```

```python
tag_elements = nltk.pos_tag(words)
```

```python
nltk.download('maxent_ne_chunker_tab')
```

```python
print(nltk.ne_chunk(tag_elements))
```
