## N-Grams

```python
import pandas as pd
# Each row is wrapped in quotes, so read it as one column and split on the first tab
messages = pd.read_csv('SMSSpamCollection.csv', header=None, names=['raw'])

messages[['label', 'message']] = messages['raw'].str.split('\t', n=1, expand=True)

messages = messages.drop(columns='raw')
```

```python
## Data Cleaning and Pre-processing
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
corpus = []
for i in range(0, len(messages)):
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if not word in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
```

```python
## Create a Bag of Words

from sklearn.feature_extraction.text import CountVectorizer
# for binary BOW enable binary=True
cv = CountVectorizer(max_features=2500, binary=True)
```

```python
X = cv.fit_transform(corpus).toarray()
```

```python
cv.vocabulary_
```

```python
## Ngram using BoW

from sklearn.feature_extraction.text import CountVectorizer
cv= CountVectorizer(max_features=200, binary=True, ngram_range=(2,3))
X = cv.fit_transform(corpus).toarray()
```
