## Stemming

```python
# Classification Problem
## check if comments on product is a positve review or negative review
## Reviews ---> eating, eat, eaten ---> eat (lemma)

words = ['eating', 'eats', 'eaten', 'writing', 'writes', 'programming', 'programs', 'history', 'finally', 'finalize']
```

#### Porter Stemmer

```python
from nltk.stem import PorterStemmer

porter_stemmer = PorterStemmer()
```

```python
for word in words:
    print(word + '----->' + porter_stemmer.stem(word))
```

```python
porter_stemmer.stem('sitting')
```

#### RegExpStemmer

```python
from nltk.stem import RegexpStemmer

reg_stemmer = RegexpStemmer(regexp='ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

#### Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snowball_stemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word + '---->' + snowball_stemmer.stem(word))
```

```python
porter_stemmer.stem('fairly'), porter_stemmer.stem('sportingly')
```

```python
snowball_stemmer.stem('fairly'), snowball_stemmer.stem('sportingly')
```

### Overall SnowBall Stemmer is better comapred to Porter Stemmer
