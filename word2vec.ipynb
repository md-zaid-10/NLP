```python
import gensim
```

```python
from gensim.models import Word2Vec, KeyedVectors
```

```python
# Using Google Word2Vec Pretrained model

import gensim.downloader as api

wv = api.load('word2vec-google-news-300')
```

```python
vec_king = wv['king']
```

```python
vec_king # This is how the word king looks like in 300 dimension
```

```python
vec_king.shape # 300 dimension
```

```python
wv['ronaldo']
```

```python
wv.most_similar('ronaldo') # this will show similar words in the corpus
```

```python
wv.similarity('football', 'swimming') # shows similarity between 2 words

# Value closer to 1 --> strongly related
# Value clsoe to 0 ---> unrelated
```

```python
vec = wv['king'] - wv['man'] + wv['woman']
```

```python
vec # 300 dim
```

```python
wv.most_similar([vec])
```
