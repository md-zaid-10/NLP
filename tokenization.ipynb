```python
corpus = '''Hello, Welcome to Mohammedzaid's NLP Github Repo.
Please do clone the repository to know more about NLP. Thank you!'''
```

```python
corpus
```

```python
## Tokenization
# Paragraph to Sentence

from nltk.tokenize import sent_tokenize
```

```python
documents = sent_tokenize(corpus)
```

```python
type(documents)
```

```python
documents
```

```python
for sentence in documents:
    print(sentence)
```

```python
### Paragraph/Sentence to words

from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus) # 's is now splitted
```

```python
# tree bank tokenizer

from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus) # full stop will not be treated as a separate word. Just for the last full stop, it will be considered separately
```
